<span style="font-size:2em; font-weight:bold;"> Grenada Rainfall Analysis: Bayesian GEV/GPD Sensitivity to Rainfall Imputationn</span>
***

This notebook extends the Grenada rainfall intensity-duration-frequency (IDF) analysis to evaluate the sensitivity of the extreme-value results to the use of imputed daily rainfall observations.

The broader IDF workflow considers two standard extreme-value analysis frameworks:

1. Generalized Pareto distribution (GPD) fitted to peaks-over-threshold (POT) exceedances.
2. Generalized extreme value (GEV) distribution fitted to annual block maxima.

The sensitivity analysis undertaken here specifically evaluates the influence of rainfall imputation on the extreme-value samples and subsequent Bayesian inference. The notebook therefore:

1. Quantifies, on a station-by-station basis, the number and proportion of annual maxima and POT exceedances derived from observed versus imputed daily rainfall observations.
2. Constructs observed-only extreme-value datasets by:
   - removing annual blocks for which the annual maximum corresponds to an imputed rainfall observation; and
   - excluding imputed POT exceedances.
3. Refits the Bayesian GEV and GPD models using the observed-only extreme-value datasets.
4. Compares the observed-only and imputation-inclusive posterior estimates, with particular emphasis on:
   - the extreme-value shape parameter, ξ; and
   - estimated rainfall return levels across the return periods considered in the main analysis.

The purpose of this analysis is to determine whether deterministic ordinary-kriging imputation materially influences the inferred upper-tail behaviour or resulting design rainfall estimates.

# 1.3 Proportion exceedances

### 1.3.1 Annual exceedances (GEV/BM)

```python
annual_max_records = []

for sta in station_ids:
    s_imp = grn_rainfall_imput[sta].dropna()

    for year, s_year in s_imp.groupby(s_imp.index.year):
        if s_year.empty:
            continue

        max_date = s_year.idxmax()
        max_value = s_year.loc[max_date]

        original_value = grn_rainfall_data.loc[max_date, sta]
        is_imputed = pd.isna(original_value)

        annual_max_records.append({
            "station": sta,
            "year": year,
            "date": max_date,
            "annual_max": max_value,
            "is_imputed": is_imputed,
        })

annual_max_df = pd.DataFrame(annual_max_records)

annual_max_summary = (
    annual_max_df.groupby("station")
    .agg(
        n_annual_max=("annual_max", "size"),
        n_observed=("is_imputed", lambda x: (~x).sum()),
        n_imputed=("is_imputed", "sum"),
    )
    .reset_index()
)

annual_max_summary["prop_imputed"] = (
    annual_max_summary["n_imputed"] / annual_max_summary["n_annual_max"]
)

annual_max_summary["pct_imputed"] = 100 * annual_max_summary["prop_imputed"]
annual_max_summary = annual_max_summary.sort_values("pct_imputed", ascending=False).reset_index(drop=True)
annual_max_summary.to_csv(r'Outputs\BAYES_MCMC_SENSITIVITY\annual_max_summary.csv', index=False)
annual_max_summary
```

### 1.3.2 POTS

```python
pot_records = []

for sta in station_ids:
    u = pots_thr.loc[pots_thr.station_id == sta, "pots_thr"].iloc[0]
    s_imp = grn_rainfall_imput[sta].dropna()

    # Threshold exceedances
    exc = s_imp[s_imp > u]

    original_values = grn_rainfall_data.loc[exc.index, sta]
    is_imputed = original_values.isna()

    for date, value, imp in zip(exc.index, exc.values, is_imputed.values):
        pot_records.append({
            "station": sta,
            "date": date,
            "threshold": float(u),
            "rainfall": float(value),
            "excess": float(value - u),
            "is_imputed": bool(imp),
        })

pot_df = pd.DataFrame(pot_records)

pot_summary = (
    pot_df.groupby("station")
    .agg(
        n_pot=("rainfall", "size"),
        n_observed=("is_imputed", lambda x: (~x).sum()),
        n_imputed=("is_imputed", "sum"),
    )
    .reset_index()
)

pot_summary["prop_imputed"] = pot_summary["n_imputed"] / pot_summary["n_pot"]
pot_summary["pct_imputed"] = 100 * pot_summary["prop_imputed"]
pot_summary = pot_summary.sort_values("pct_imputed", ascending=False).reset_index(drop=True)
pot_summary.to_csv(r'Outputs\BAYES_MCMC_SENSITIVITY\pot_summary.csv', index=False)
pot_summary
```

# 2.0 Jax/Numpyro model definitions

## 2.1 Skew_normal distribution

```python
class SkewNormalPrior(Distribution):
    arg_constraints = {
        "loc": constraints.real,
        "scale": constraints.positive,
        "alpha": constraints.real,
    }
    support = constraints.real
    has_rsample = False

    def __init__(self, loc, scale, alpha, validate_args=None):
        batch_shape = lax.broadcast_shapes(
            jnp.shape(loc), jnp.shape(scale), jnp.shape(alpha)
        )
        self.loc, self.scale, self.alpha = promote_shapes(
            loc, scale, alpha, shape=batch_shape
        )
        super().__init__(batch_shape=batch_shape, event_shape=(), validate_args=validate_args)

    def sample(self, key, sample_shape=()):
        assert is_prng_key(key)
        k1, k2 = jax.random.split(key)
        shape = sample_shape + self.batch_shape

        u0 = jax.random.normal(k1, shape=shape)
        u1 = jax.random.normal(k2, shape=shape)

        delta = self.alpha / jnp.sqrt(1.0 + self.alpha**2)
        z = delta * jnp.abs(u0) + jnp.sqrt(1.0 - delta**2) * u1

        return self.loc + self.scale * z

    def log_prob(self, value):
        z = (value - self.loc) / self.scale
        base = dist.Normal(0.0, 1.0).log_prob(z)
        skew = log_ndtr(self.alpha * z)
        return jnp.log(2.0) - jnp.log(self.scale) + base + skew
```

## 2.2 Generalized Pareto Distribution (GPD)

```python
GPD_EPS = 1e-6

def _gpd_logpdf(y, sigma, xi):
    y, sigma, xi = promote_shapes(y, sigma, xi)
    t = 1.0 + xi * (y / sigma)

    # validity: y >= 0, sigma > 0, support condition
    valid = (y >= 0.0) & (sigma > 0.0) & (t > 0.0)

    # xi != 0
    logpdf_xi = -jnp.log(sigma) - (1.0 / xi + 1.0) * jnp.log(t)

    # xi -> 0 (Exponential limit)
    logpdf_exp = -jnp.log(sigma) - y / sigma

    logpdf = jnp.where(jnp.abs(xi) < GPD_EPS, logpdf_exp, logpdf_xi)
    return jnp.where(valid, logpdf, -jnp.inf)


def _gpd_icdf(p, sigma, xi):
    p, sigma, xi = promote_shapes(p, sigma, xi)
    p = jnp.clip(p, 1e-12, 1.0 - 1e-12)

    y_xi = sigma * ((1.0 - p) ** (-xi) - 1.0) / xi
    y_exp = -sigma * jnp.log(1.0 - p)

    return jnp.where(jnp.abs(xi) < GPD_EPS, y_exp, y_xi)


class GenPareto(Distribution):
    arg_constraints = {
        "sigma": constraints.positive,
        "xi": constraints.real,
    }
    support = constraints.nonnegative
    has_rsample = False

    def __init__(self, sigma, xi, validate_args=None):
        batch_shape = lax.broadcast_shapes(jnp.shape(sigma), jnp.shape(xi))
        self.sigma, self.xi = promote_shapes(sigma, xi, shape=batch_shape)
        super().__init__(batch_shape=batch_shape, event_shape=(), validate_args=validate_args)

    def sample(self, key, sample_shape=()):
        assert is_prng_key(key)
        u = jax.random.uniform(
            key,
            shape=sample_shape + self.batch_shape,
            minval=1e-12,
            maxval=1.0 - 1e-12,
        )
        return _gpd_icdf(u, self.sigma, self.xi)

    def log_prob(self, value):
        return _gpd_logpdf(value, self.sigma, self.xi)

def model_gpd(y, sigma_prior=(10.0,), xi_prior=(0.0, 0.15)):
    sigma = numpyro.sample("sigma", dist.LogNormal(jnp.log(sigma_prior[0]), 0.5))
    xi = numpyro.sample("xi", dist.TruncatedNormal(
        low=-0.5, loc=xi_prior[0], scale=xi_prior[1], high=0.5
    ))
    numpyro.sample("obs", GenPareto(sigma, xi), obs=y)

def model_gpd(y):
    log_sigma = numpyro.sample("log_sigma",dist.StudentT(df=95.92, loc=2.83, scale=0.32))
    sigma = jnp.exp(log_sigma)
    xi = numpyro.sample("xi", SkewNormalPrior(loc=0.02, scale=0.25, alpha=1.0))
    numpyro.sample("obs", GenPareto(sigma, xi), obs=y)
```

## 2.3 General Extreme Value (GEV) distribution

```python
GEV_EPS = 1e-6

def _gev_logpdf(x, mu, sigma, xi):
    x, mu, sigma, xi = promote_shapes(x, mu, sigma, xi)
    z = (x - mu) / sigma
    t = 1.0 + xi * z

    # validity: sigma > 0 and support condition
    valid = (sigma > 0.0) & (t > 0.0)

    # xi != 0
    logpdf_xi = (
        -jnp.log(sigma)
        - (1.0 + 1.0 / xi) * jnp.log(t)
        - t ** (-1.0 / xi)
    )

    # xi -> 0 (Gumbel limit)
    logpdf_gumbel = -jnp.log(sigma) - z - jnp.exp(-z)

    logpdf = jnp.where(jnp.abs(xi) < GEV_EPS, logpdf_gumbel, logpdf_xi)
    return jnp.where(valid, logpdf, -jnp.inf)


def _gev_icdf(p, mu, sigma, xi):
    p, mu, sigma, xi = promote_shapes(p, mu, sigma, xi)
    p = jnp.clip(p, 1e-12, 1.0 - 1e-12)

    term = -jnp.log(p)

    x_xi = mu + sigma * ((term ** (-xi) - 1.0) / xi)
    x_gumbel = mu - sigma * jnp.log(term)

    return jnp.where(jnp.abs(xi) < GEV_EPS, x_gumbel, x_xi)


class GenExtreme(Distribution):
    arg_constraints = {
        "mu": constraints.real,
        "sigma": constraints.positive,
        "xi": constraints.real,
    }
    support = constraints.real
    has_rsample = False

    def __init__(self, mu, sigma, xi, validate_args=None):
        batch_shape = lax.broadcast_shapes(
            jnp.shape(mu), jnp.shape(sigma), jnp.shape(xi)
        )
        self.mu, self.sigma, self.xi = promote_shapes(
            mu, sigma, xi, shape=batch_shape
        )
        super().__init__(batch_shape=batch_shape, event_shape=(), validate_args=validate_args)

    def sample(self, key, sample_shape=()):
        assert is_prng_key(key)
        u = jax.random.uniform(
            key,
            shape=sample_shape + self.batch_shape,
            minval=1e-12,
            maxval=1.0 - 1e-12,
        )
        return _gev_icdf(u, self.mu, self.sigma, self.xi)

    def log_prob(self, value):
        return _gev_logpdf(value, self.mu, self.sigma, self.xi)

def model_gev(z):
    mu = numpyro.sample("mu", dist.InverseGamma(9.2, 528.1))
    log_sigma = numpyro.sample("log_sigma",dist.StudentT(df=95.92, loc=2.83, scale=0.45))
    sigma = jnp.exp(log_sigma)
    xi = numpyro.sample("xi", SkewNormalPrior(loc=0.02, scale=0.25, alpha=1.0))
    numpyro.sample("obs", GenExtreme(mu, sigma, xi), obs=z)
```

# 4.0 Bayesian MCMC fitting

## 4.1 Run model: GPD

```python
results_gpd_obs = {}
predictions_gpd_obs = {}

for i, sta in enumerate(station_ids):
    rng_key = jax.random.PRNGKey(1000 + i)

    u = pots_thr.loc[pots_thr.station_id == sta, "pots_thr"].iloc[0]
    s_imp = grn_rainfall_imput[sta].dropna()

    # Same declustered POT sample as the imputation-inclusive analysis
    pot = get_extremes(
        ts=s_imp,
        method="POT",
        extremes_type="high",
        threshold=float(u),
        r="24h",
    )

    # Retain only declustered peaks that were genuinely observed
    observed_mask = grn_rainfall_data.loc[pot.index, sta].notna()
    pot_obs = pot.loc[observed_mask]

    y = jnp.asarray(pot_obs.to_numpy() - u)

    kernel = NUTS(model_gpd, target_accept_prob=0.98)
    mcmc = MCMC(kernel, num_warmup=2000, num_samples=1000, num_chains=4)
    mcmc.run(rng_key, y)

    samples = mcmc.get_samples()

    predictive = Predictive(model_gpd, posterior_samples=samples)
    pp = predictive(jax.random.PRNGKey(12345), y)

    idata = az.from_numpyro(
        mcmc,
        posterior_predictive=pp,
        log_likelihood=True,
    )

    results_gpd_obs[sta] = {
        "threshold": float(u),
        "n_exceedances": len(y),
        "mcmc": mcmc,
        "samples": samples,
    }

    predictions_gpd_obs[sta] = idata

    clear_output(wait=True)

# export trace plots
for sta in station_ids:
    fig = plt.figure(figsize=(10, 6), dpi=200)
    az.plot_trace(predictions_gpd_obs[sta])
    plt.tight_layout()
    plt.savefig(
        rf"Outputs\BAYES_MCMC_SENSITIVITY\BAYES_MCMC_TRACE\GPD_{sta}.png",
        dpi=200,
        bbox_inches="tight"
    )
    plt.close('all')
```

## 4.2 GEV

```python
results_gev_obs = {}
predictions_gev_obs = {}

for i, sta in enumerate(station_ids):
    rng_key = jax.random.PRNGKey(1000 + i)

    obs = (
        annual_max_df.loc[
            (annual_max_df["station"] == sta) &
            (~annual_max_df["is_imputed"]),
            "annual_max"
        ]
        .to_numpy()
    )

    kernel = NUTS(model_gev, target_accept_prob=0.98)
    mcmc = MCMC(kernel, num_warmup=2000, num_samples=1000, num_chains=4)
    mcmc.run(rng_key, obs)

    samples = mcmc.get_samples()

    predictive = Predictive(model_gev, posterior_samples=samples)
    pp = predictive(jax.random.PRNGKey(12345), obs)

    idata = az.from_numpyro(
        mcmc,
        posterior_predictive=pp,
        log_likelihood=True,
    )

    results_gev_obs[sta] = {
        "n_blocks": len(obs),
        "mcmc": mcmc,
        "samples": samples,
    }

    predictions_gev_obs[sta] = idata

    clear_output(wait=True)

# export trace plots
for sta in station_ids:
    fig = plt.figure(figsize=(10, 6), dpi=200)
    az.plot_trace(predictions_gev_obs[sta])
    plt.tight_layout()
    plt.savefig(
        rf"Outputs\BAYES_MCMC_SENSITIVITY\BAYES_MCMC_TRACE\GEV_{sta}.png",
        dpi=200,
        bbox_inches="tight"
    )
    plt.close('all')
```

```python
rows = []

for sta in station_ids:

    gev_post = pd.read_csv(post_dir / rf'BAYES_MCMC_PARAMS\{sta}_GEV_samples.csv')
    gpd_post = pd.read_csv(post_dir / rf'BAYES_MCMC_PARAMS\{sta}_GPD_samples.csv')

    # GPD needs threshold and lambda
    u = pots_thr.loc[pots_thr.station_id == sta, "pots_thr"].iloc[0]
    lam = exceedance_rate_per_year(grn_rainfall_imput[sta], u)

    gev_row = {"station": sta, "dist": "GEV"}
    gpd_row = {"station": sta, "dist": "GPD"}

    # Summaries
    gev_summary = gev_return_levels_from_posterior(T, gev_post, quantiles=(0.05, 0.95))
    gpd_summary = gpd_return_levels_from_posterior(T, gpd_post, u=u, lam_per_year=lam, quantiles=(0.05, 0.95))

    for T_i in T:
        label = f"{T_i:g}"

        mask_gev = gev_summary["T"].astype(float) == float(T_i)
        row_gev = gev_summary.loc[mask_gev].iloc[0]

        gev_row[f"TR_{label}_mean"] = row_gev["mean"]
        gev_row[f"TR_{label}_std"]  = row_gev["std"]
        # gev_row[f"TR_{label}_5"]    = row_gev["q05"]
        # gev_row[f"TR_{label}_95"]   = row_gev["q95"]

        mask_gpd = gpd_summary["T"].astype(float) == float(T_i)
        row_gpd = gpd_summary.loc[mask_gpd].iloc[0]

        gpd_row[f"TR_{label}_mean"] = row_gpd["mean"]
        gpd_row[f"TR_{label}_std"]  = row_gpd["std"]
        # gpd_row[f"TR_{label}_5"]    = row_gpd["q05"]
        # gpd_row[f"TR_{label}_95"]   = row_gpd["q95"]

    rows.append(gev_row)
    rows.append(gpd_row)

geva_tr_post_obs = pd.DataFrame(rows)
```
